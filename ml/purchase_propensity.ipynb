# Purchase Propensity

## Objectif
Ce notebook présente les variables permettant de construire un modèle de propension à l’achat.

L’objectif serait d’estimer la probabilité qu’un segment ou une région génère une hausse des ventes sur une période donnée, à partir des performances passées.


## Features utilisées

Les variables sélectionnées proviennent du data product marketing :

- Nombre de transactions
- Ventes totales
- Panier moyen
- Budget marketing
- Reach
- Taux de conversion moyen
- Ventes du mois précédent 
- Variation des ventes MoM

Ces variables sont adaptées à des modèles supervisés (logistic regression, random forest, gradient boosting).


```python
SELECT
  month,
  region,
  nb_transactions,
  total_sales,
  avg_basket,
  total_budget,
  total_reach,
  avg_conversion_rate,
  sales_prev_month,
  sales_delta_mom
FROM ANYCOMPANY_LAB.ANALYTICS.MARKETING_DATAPRODUCT;
```
