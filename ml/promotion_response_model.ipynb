# Promotion Response Model 

## Objectif
Ce notebook illustre comment mesurer l’impact des promotions sur les ventes et leur évolution dans le temps.

L’objectif serait de prédire l’efficacité d’une promotion en fonction de son intensité et de sa fréquence.


## Features utilisées

Les variables exploitées incluent :

- Nombre de promotions
- Remise moyenne
- Remise maximale
- Ventes totales
- Variation des ventes mensuelle

Ces variables sont adaptées à des modèles de régression ou à des approches causales.
